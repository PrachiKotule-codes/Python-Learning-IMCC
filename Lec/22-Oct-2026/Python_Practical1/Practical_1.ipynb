```python
# Question 1: Read student names from a file

file = open("students.txt", "r")

names = file.readlines()

file.close()

names = [name.strip() for name in names]

print(names)
```

```python

# Question 2: Print names longer than 5 characters
file = open("students.txt", "r")
names = file.readlines()
file.close()


for name in names:
    if len(name) > 5:
        print(name)
```

```python

# Question 3: Count names starting with a vowel from a file

file = open("students.txt", "r")
names = file.readlines()
file.close()

count = 0

for name in names:
    name = name.strip()

    if name[0].lower() in "aeiou":
        count = count + 1

print("Number of names starting with a vowel:", count)
```

```python

# Question 4: Save long names into a new file
file = open("students.txt", "r")
names = file.readlines()
file.close()

long_names = []

for name in names:
    if len(name) > 5:
        long_names.append(name)

file = open("long_names.txt", "w")

for name in long_names:
    file.write(name + "\n")

file.close()

print("Names saved successfully!")
```

```python

# Question 5: Sort names alphabetically
file = open("students.txt", "r")
names = file.readlines()
file.close()

names.sort()

print("Names Are in alphabetical Order")

for name in names:
    print(name)
```

```python
# Question 7: Count word frequency in names (handle duplicates) and store

file = open("students.txt", "r")
names = file.readlines()
file.close()

frequency = {}

for name in names:
    if name in frequency:
        frequency[name] = frequency[name] + 1
    else:
        frequency[name] = 1

print(frequency)
```

```python

# Question 8: String Analysis

for name in names:
    print("Name:", name)
    print("Length:", len(name))
    print("First character:", name[0])
    print("Last character:", name[-1])

    if name.lower() == name[::-1].lower():
        print("Palindrome: Yes")
    else:
        print("Palindrome: No")

```

```python

# Question 9: Remove duplicates and sort by length

file = open("students.txt", "r")
names_list = file.readlines()
file.close()

names_list = [name.strip() for name in names_list]

unique_names = list(set(names_list))

unique_names.sort(key=len)

file = open("processed_names.txt", "w")

for name in unique_names:
    file.write(name + "\n")

file.close()

print("Names processed and saved successfully!")
print(unique_names)
```

```python

# Question 10: Group names by starting letter

file = open("students.txt", "r")
names_list = file.readlines()
file.close()

groups = {}

for name in names_list:
    name = name.strip()

    if name:
        letter = name[0].upper()

        if letter not in groups:
            groups[letter] = []

        groups[letter].append(name)

for letter in groups:
    filename = letter + "_names.txt"

    file = open(filename, "w")

    for name in groups[letter]:
        file.write(name + "\n")

    file.close()

print("Names grouped into separate files successfully!")
print(groups)
```
