```python
# Q2: Customer Data Cleaning

# List containing customer information

lines = [
    ' asha VERMA ; ASHA.V@Example.COM ; +91 98765-43210 ;pune ',
    'ravi kumar;ravi@@corp.in;9123456780;Mumbai',
    'meera;meera@mail;9123456780;Delhi',
    'john doe; john@x.org ;12345; Goa',
    'Neha Rao;neha.rao@firm.co.in;919876501234;nashik',
    'amit;amit@site.com;9876543210'
]

print("Customer records loaded successfully!")
```

```python
# Function to clean and validate customer records

def clean_customer(raw):

    # Split the record using semicolon
    data = raw.split(';')

    # Check whether there are exactly 4 fields
    if len(data) != 4:
        return None

    # Remove extra spaces from all fields
    name = data[0].strip()
    email = data[1].strip().lower()
    phone = data[2].strip()
    city = data[3].strip()

    # -------------------------------
    # 1. Validate Name
    # -------------------------------

    # Convert name into Title Case
    name = name.title()

    # Split name into individual words
    words = name.split()

    # Name should not be empty
    if len(words) == 0:
        return None

    # Every word should contain only letters
    for word in words:
        if not word.isalpha():
            return None

    # -------------------------------
    # 2. Validate City
    # -------------------------------

    # City should not be empty
    if city == "":
        return None

    # Convert city into Title Case
    city = city.title()

    # -------------------------------
    # 3. Validate Email
    # -------------------------------

    # Email should contain exactly one @
    if email.count('@') != 1:
        return None

    # Split email into username and domain
    username, domain = email.split('@')

    # Username should not be empty
    if username == "":
        return None

    # Domain should contain a dot
    if '.' not in domain:
        return None

    # Dot should not be first or last in domain
    if domain.startswith('.') or domain.endswith('.'):
        return None

    # -------------------------------
    # 4. Validate Phone Number
    # -------------------------------

    # Remove spaces and hyphens
    phone = phone.replace(" ", "")
    phone = phone.replace("-", "")

    # Remove +91 if present
    if phone.startswith("+91"):
        phone = phone[3:]

    # Remove 91 if it is a 12-digit number
    elif len(phone) == 12 and phone.startswith("91"):
        phone = phone[2:]

    # Check phone number length
    if len(phone) != 10:
        return None

    # Check whether phone contains only digits
    if not phone.isdigit():
        return None

    # Phone number must start with 6, 7, 8 or 9
    if phone[0] not in "6789":
        return None

    # -------------------------------
    # 5. Return Valid Customer
    # -------------------------------

    customer = {
        'name': name,
        'email': email,
        'phone': phone,
        'city': city
    }

    return customer
```

```python
# Count invalid customer records
rejected = 0

# Check each customer record
for raw in lines:

    # Call the cleaning function
    customer = clean_customer(raw)

    # If record is valid, print it
    if customer is not None:
        print(customer)

    # Otherwise increase rejected count
    else:
        rejected = rejected + 1

# Display total rejected records
print("Rejected:", rejected)
```
